```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from google.colab import files

# 1. URL target: Berita/Artikel tentang TikTok Indonesia dari CNN Indonesia
url = "https://www.cnnindonesia.com/tag/tiktok"

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

print("Sedang mengambil data dari web...")
response = requests.get(url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')

    # Mencari elemen artikel
    articles = soup.find_all('article')

    data_list = []

    for article in articles:
        # Mengambil judul berita
        title_tag = article.find('h2') or article.find('h3')
        # Mengambil link berita
        link_tag = article.find('a')
        # Mengambil kategori/waktu jika ada
        date_tag = article.find('span', class_='date')

        if title_tag and link_tag:
            judul = title_tag.text.strip()
            link = link_tag['href']
            tanggal = date_tag.text.strip() if date_tag else "N/A"

            data_list.append({
                'Judul Berita': judul,
                'Tanggal': tanggal,
                'Link': link
            })

    # 2. Mengubah data menjadi DataFrame Pandas
    df = pd.DataFrame(data_list)

    # Tampilkan preview 5 data teratas di Colab
    print(f"\nBerhasil mengambil {len(df)} data tentang TikTok Indonesia!\n")
    display(df.head())

    # 3. Menyimpan ke file CSV
    nama_file = "data_tiktok_indonesia.csv"
    df.to_csv(nama_file, index=False, encoding='utf-8-sig')
    print(f"Data disimpan ke {nama_file}")

    # 4. Mengunduh file CSV otomatis ke komputer Anda
    files.download(nama_file)

else:
    print(f"Gagal mengakses web. Status code: {response.status_code}")
```
